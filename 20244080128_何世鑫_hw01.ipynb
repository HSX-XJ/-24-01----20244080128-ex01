```python
# ==========================================
# 任务1：热身：定义简单函数
# ==========================================

# 1. 定义 say_hi 函数，不接受参数，打印 "Hello, world!"
def say_hi():
    print("Hello, world!")

# 2. 定义 goat_pad 函数，接受一个字符串，前后加上 "goat" 并打印
def goat_pad(s):
    print("goat" + s + "goat")

# 3. 定义 print_n 函数，接受字符串 s 和整数 n，将字符串打印 n 次，每次单独一行
def print_n(s, n):
    for _ in range(n):
        print(s)

# ==========================================
# 以下是测试代码，用于验证函数是否正确
# ==========================================
print("--- 测试 say_hi ---")
say_hi()

print("\n--- 测试 goat_pad ---")
goat_pad("hello")

print("\n--- 测试 print_n ---")
print_n("Python", 3)
```

```python
# ==========================================
# 任务2：欧几里得算法
# ==========================================

# 1. 实现 gcd 函数，接受两个整数并返回它们的最大公约数
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return abs(m)

# ==========================================
# 2. 评估指定数字对的 GCD
# ==========================================
print("--- 测试 GCD ---")
print(f"gcd(2018, 2019) = {gcd(2018, 2019)}")
print(f"gcd(1200, 300)  = {gcd(1200, 300)}")
print(f"gcd(5040, 60)   = {gcd(5040, 60)}")

# 补充测试负数情况，用于回答第3小问
print(f"\ngcd(-2018, 2019) = {gcd(-2018, 2019)}")
print(f"gcd(-1200, -300) = {gcd(-1200, -300)}")


# ==========================================
# 3. 文字解答：如果一个或两个参数是负数，函数会做什么？这种行为有意义吗？
# ==========================================
print("\n--- 第3问解答 ---")
answer = """
如果参数中包含负数：
- 如果函数里没有使用 abs()，由于 Python 的取模运算 (m % n) 结果符号跟随除数，
  会导致返回的结果可能是负数。例如 gcd(-2018, 2019) 可能会返回 -1。
- 数学上，最大公约数 (GCD) 定义为非负整数。因此，在函数返回时使用了 abs(m)，
  确保无论输入是正数还是负数，都会返回有意义且正确的非负公约数。

若两个参数都为 0，函数会直接跳过 while 循环并返回 0，这也符合数学上 gcd(0, 0) = 0 的约定。
"""
print(answer)
```

```python
# ==========================================
# 任务3：近似欧拉数 e
# ==========================================
import math

# 1. 定义 euler_limit 函数：取 x = n 来近似 e，即 (1 + 1/n)^n
def euler_limit(n):
    return (1 + 1 / n) ** n

# 2. 定义 euler_infinite_sum 函数：取公式(1)前 n 项之和。
# 注意：公式从 k=0 开始，我们的函数 n 从 1 开始计数。
# 所以前 n 项意味着 k 从 0 到 n-1。
def euler_infinite_sum(n):
    total = 0.0
    factorial = 1  # 0! 的阶乘是 1
    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1 / factorial
    return total

# 3. 定义 euler_approx 函数：累加直到误差在 epsilon 之内
def euler_approx(epsilon):
    true_e = math.exp(1)  # 使用 math 模块获取真实 e 值
    total = 0.0
    factorial = 1
    k = 0
    # 当当前和与真实 e 的差值大于等于 epsilon 时，继续循环累加
    while abs(total - true_e) >= epsilon:
        if k > 0:
            factorial *= k
        total += 1 / factorial
        k += 1
    return total

# 4. 定义 print_euler_sum_table 函数：打印 k 从 1 到 n 的 euler_infinite_sum(k)
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))


# ==========================================
# 以下是测试代码，用于验证函数是否正确
# ==========================================
print("--- 测试 euler_limit ---")
print(f"euler_limit(1000) = {euler_limit(1000)}")

print("\n--- 测试 euler_infinite_sum ---")
print(f"euler_infinite_sum(4) = {euler_infinite_sum(4)}") # 预期约 2.667
print(f"euler_infinite_sum(1) = {euler_infinite_sum(1)}") # 预期 1.0
print(f"euler_infinite_sum(0) = {euler_infinite_sum(0)}") # 预期 0.0

print("\n--- 测试 euler_approx ---")
print(f"euler_approx(1e-6) = {euler_approx(1e-6)}")
print(f"真实 e 值 (math.exp(1)) = {math.exp(1)}")

print("\n--- 测试 print_euler_sum_table (n=5) ---")
print_euler_sum_table(5)
```

```python
# ==========================================
# 任务4：测试整数属性
# ==========================================

# 1. 定义 is_power_of_2 函数：判断正整数 n 是否为 2 的幂
def is_power_of_2(n):
    if n <= 0:
        return False
    # 不断除以 2，直到不能整除为止
    while n % 2 == 0:
        n //= 2
    # 如果最后剩下的数字是 1，说明是 2 的幂
    return n == 1

# 2. 推广到 is_power 函数：判断正整数 n 是否为 b 的幂
def is_power(b, n):
    # 处理特殊情况：如果底数或目标数小于等于 0，直接返回 False
    if b <= 0 or n <= 0:
        return False
    # 如果底数是 1，只有 n 也是 1 时才成立（1^p = 1）
    if b == 1:
        return n == 1
    
    # 不断除以 b，直到不能整除为止
    while n % b == 0:
        n //= b
    # 如果最后剩下的数字是 1，说明是 b 的幂
    return n == 1


# ==========================================
# 以下是测试代码，用于验证函数是否正确
# ==========================================
print("--- 测试 is_power_of_2 ---")
print(f"is_power_of_2(1)    = {is_power_of_2(1)}")    # True (2^0)
print(f"is_power_of_2(2)    = {is_power_of_2(2)}")    # True
print(f"is_power_of_2(8)    = {is_power_of_2(8)}")    # True
print(f"is_power_of_2(6)    = {is_power_of_2(6)}")    # False
print(f"is_power_of_2(1024) = {is_power_of_2(1024)}") # True

print("\n--- 测试 is_power ---")
print(f"is_power(3, 27) = {is_power(3, 27)}")  # True (3^3)
print(f"is_power(3, 28) = {is_power(3, 28)}")  # False
print(f"is_power(2, 64) = {is_power(2, 64)}")  # True (2^6)
print(f"is_power(5, 125) = {is_power(5, 125)}") # True (5^3)
```
