```python
print("实验二:1")
def score_level(score):
    if  type(score) == float or type(score) == int:
        if score > 100:
           raise ValueError(score)
        elif score >= 90:
            print("优秀")
        elif score >= 80:
            print("良好")
        elif score >= 60:
            print("及格")
        elif score >= 0:
            print("不合格")
        else:
            raise ValueError(score)
    else :
        raise TypeError(score)


for s in [98, 102, "ss", -12]:
    try:
        score_level(s)
    except Exception as e:
        print(f"输入 {s!r} 触发 {type(e).__name__}: {e}")

print("实验二:2")
# ===== 方法一：for 循环 =====
total_for = 0
for i in range(1, 101):      # range(1, 101) 生成 1,2,3,...,100
    total_for += i
print("for  循环求和结果:", total_for)

# ===== 方法二：while 循环 =====
total_while = 0
i = 1                        # 初始化循环变量
while i <= 100:
    total_while += i
    i += 1                   # 别忘了自增，否则死循环
print("while 循环求和结果:", total_while)
#验证
n = 100
formula = n * (n + 1) // 2
print("公式 n(n+1)/2 结果:", formula)
print("与循环结果一致:", formula == total_for == total_while)

print("实验二:3")
#递归
def factorial_recursive(number_1:int):
    if number_1 == 0:
        return 1
    else:
        return number_1*factorial_recursive(number_1 - 1)

print(factorial_recursive(5))   

#迭代
def for_iterative(number_2:int):
    if number_2 == 0 or number_2 == 1:
        return 1 
    else:    
        num_all = 1
        for n in range(1,number_2 + 1):
            num_all = n*num_all
        return num_all

print(for_iterative(5))    

#比较
# ===== 比较 n = 0、1、5、10 时的结果 =====
for n in [0, 1, 5, 10]:
    r_rec = factorial_recursive(n)      # 递归版
    r_ite = for_iterative(n)            # 迭代版（你原来的名字）
    same = (r_rec == r_ite)             # 两者是否相等
    print(f"n={n:>2}: 递归={r_rec}, 迭代={r_ite}, 相等={same}")

print("实验二:4")
def summary(values:list):
    tuple_1 = (min(values), max(values), sum(values)/len(values))
    return tuple_1
list_text = list(range(1,6))
tuple_text_min, tuple_text_max, tuple_text_ = summary(list_text)
print(tuple_text_min,tuple_text_max,tuple_text_)

print("实验二:5")

def sum_all(*args):
    return sum(args)
# ===== 测试 =====
print("无参数:", sum_all())               # 期望 0
print("单个参数:", sum_all(5))            # 期望 5
print("多个参数:", sum_all(1, 2, 3, 4, 5)) # 期望 15
```

```python
print("实验四")
print("1")
import os
# 1. 查看当前工作目录
cwd = os.getcwd()
print("当前工作目录:", cwd)

# 2. 构造数据目录的相对路径
data_dir = os.path.join(cwd, "exp01_data")
print("数据目录:", data_dir)

# 3. 创建目录（如果已存在不报错）
os.makedirs(data_dir, exist_ok=True)
print("目录已就绪")

# 要写入的内容
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete"""

# 1. 写文件
file_path = os.path.join(data_dir, "students.txt")
with open(file_path, "w", encoding="utf-8") as f:
    f.write(content)

print("写入完成")

# 2. 读文件并跳过表头
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

# 跳过第一行表头，逐行解析
for i, line in enumerate(lines[1:], start=2):   # start=2 表示从文件第 2 行开始计数
    line = line.strip()
    if not line:
        continue
    parts = line.split(",")
    print(f"第 {i} 行:", parts)

def parse_student(line):
    """
    解析一行学生记录。
    合法返回 {"name": str, "score": int}；
    非法抛出 ValueError，并说明原因。
    """
    # 1. 去掉首尾空白（包含换行符）
    line = line.strip()

    # 2. 按逗号切分
    parts = line.split(",")

    # 3. 检查字段数量
    if len(parts) != 2:
        raise ValueError(f"字段数量错误：期望 2 个，实际 {len(parts)} 个")

    # 4. 解包
    name, score_str = parts

    # 5. 检查姓名是否为空
    if not name:
        raise ValueError("姓名不能为空")

    # 6. 尝试转换成绩为整数
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是数字：'{score_str}'")

    # 7. 检查成绩范围
    if score < 0 or score > 100:
        raise ValueError(f"成绩超出范围：{score}")

    # 8. 返回结果
    return {"name": name, "score": score}

for line in lines:
    try:
        result = parse_student(line)
        print(f"✅ 合法: {line!r} → {result}")
    except ValueError as e:
        print(f"❌ 无效: {line!r} → 错误: {e}")

# 1. 准备两个空列表
valid_records = []   # 存放解析成功的字典，如 {"name": "Alice", "score": 85}
error_records = []   # 存放错误信息，格式为 (行号, 错误原因)

# 2. 读取文件（复用任务 2 的 file_path 和读取方式）
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

# 3. 遍历数据行（跳过表头，行号从 2 开始）
for i, line in enumerate(lines[1:], start=2):
    # 去除行首尾的空白字符（包括换行符）
    line = line.strip()
    
    # 如果是空行，直接跳过，不处理
    if not line:
        continue
        
    try:
        # 【尝试解析】调用我们上个任务写好的函数
        record = parse_student(line)
        # 如果解析成功没有报错，就把结果存入 valid_records 列表
        valid_records.append(record)
        
    except ValueError as e:
        # 【捕获异常】如果解析失败，捕获 ValueError
        # 将"当前行号 i"和"错误信息 str(e)"打包成元组，存入 error_records
        error_records.append((i, str(e)))

# ==========================================
# 4. 输出统计结果
# ==========================================
print("=" * 30)
print(f"有效记录数: {len(valid_records)}")
print(f"无效记录数: {len(error_records)}")

print("\n--- 错误明细 ---")
# 解包元组，分别打印行号和原因
for lineno, msg in error_records:
    print(f"第 {lineno} 行: {msg}")

print("\n--- 有效记录 ---")
for r in valid_records:
    print(r)

# ==========================================
# 任务 5：将统计结果写入 report.txt，并演示 w 和 a 模式
# ==========================================

# 1. 数据统计（基于任务 4 的 valid_records 和 error_records）
scores = [r["score"] for r in valid_records]

valid_count = len(valid_records)
error_count = len(error_records)

# 防止列表为空导致除以 0 报错
if valid_count > 0:
    avg = sum(scores) / valid_count
    passed_count = sum(1 for s in scores if s >= 60)
    pass_rate = passed_count / valid_count
else:
    avg = 0
    pass_rate = 0

# 2. 写入文件（使用 "w" 模式：覆盖）
report_path = os.path.join(data_dir, "report.txt")

with open(report_path, "w", encoding="utf-8") as f:
    f.write("========== 成绩统计报告 ==========\n")
    f.write(f"有效记录数: {valid_count}\n")
    f.write(f"无效记录数: {error_count}\n")
    f.write(f"平均分: {avg:.2f}\n")
    f.write(f"及格率: {pass_rate:.2%}\n")
    f.write("==================================\n")

print("【第一次读取】w 模式写入后，文件内容为：")
with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())

# 3. 演示追加（使用 "a" 模式：追加）
# 我们故意用 a 模式再写一次，看看会不会覆盖上面的内容
with open(report_path, "a", encoding="utf-8") as f:
    f.write("\n【附加信息】此报告由系统使用 a 模式追加生成。\n")

print("【第二次读取】a 模式追加后，文件内容为：")
with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())

import pickle

# 1. 构造 pkl 文件路径
pkl_path = os.path.join(data_dir, "students.pkl")

# ==========================================
# 写入：把 valid_records 存入 .pkl 文件
# ==========================================
with open(pkl_path, "wb") as f:       # 注意是 "wb"（二进制写）
    pickle.dump(valid_records, f)     # 对象在前，文件在后
print("pickle 序列化完成")

# ==========================================
# 读取：从 .pkl 文件恢复数据
# ==========================================
with open(pkl_path, "rb") as f:       # 注意是 "rb"（二进制读）
    loaded_records = pickle.load(f)
print("pickle 反序列化完成")

# ==========================================
# 验证：恢复的数据是否与原数据相等
# ==========================================
print("\n原数据长度:", len(valid_records))
print("恢复数据长度:", len(loaded_records))
print("内容是否完全一致:", loaded_records == valid_records)

# 任务 7：测试导入模块
from student_utils import parse_student

print("--- 测试一个有效输入 ---")
valid_result = parse_student("Alice,85")
print("测试结果:", valid_result)

print("\n--- 测试一个无效输入 ---")
try:
    parse_student("Grace,abc")
except ValueError as e:
    print("成功捕获异常:", e)

# 任务 8：尝试读取一个不存在的文件，并捕获 FileNotFoundError
nonexistent_path = os.path.join(data_dir, "this_file_does_not_exist.txt")
print(f"尝试读取的路径: {nonexistent_path}")

try:
    with open(nonexistent_path, "r", encoding="utf-8") as f:
        content = f.read()
    print("读取成功（这行不会被执行到）")
except FileNotFoundError as e:
    print("\n[捕获到异常] 文件不存在！")
    print(f"错误详情: {e}")
    print("说明：程序没有崩溃，而是平稳地处理了这次读取失败。")
```

```python
print("实验五")
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        # 成绩大于等于 60 视为及格
        return self.score >= 60

    def __str__(self):
        # print(对象) 时会自动调用此方法
        return f"姓名: {self.name}, 成绩: {self.score}"

# 简单测试
s = Student("Alice", 85)
print(s)
print("是否及格:", s.is_passed())

# 基于你实验 4 收集到的 valid_records（包含 6 个有效学生）
students = []

for record in valid_records:
    # record 形如 {"name": "Alice", "score": 85}
    stu = Student(record["name"], record["score"])
    students.append(stu)
    
    status = "及格" if stu.is_passed() else "不及格"
    print(f"{stu} -> {status}")

class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        # 用 super() 调用父类的初始化，把 name 和 score 传上去
        super().__init__(name, score)
        # 增加子类独有的属性
        self.research_topic = research_topic

    def __str__(self):
        # 重写 __str__，包含研究方向信息
        return f"研究生姓名: {self.name}, 成绩: {self.score}, 研究方向: {self.research_topic}"

# 测试子类
gs = GraduateStudent("Bob", 88, "人工智能")
print(gs)
print("是否及格:", gs.is_passed())  # 继承了父类的方法

import copy

print("-" * 20, "1. 直接赋值", "-" * 20)
# 新建原始对象
orig_1 = {"name": "Alice", "scores": [85, 90]}
alias = orig_1               # 直接赋值（别名）
alias["scores"].append(95)   # 修改内部列表
print("原对象 (orig_1):", orig_1)
print("别名   (alias) :", alias)
# 结果：原对象内部的列表也会变成 [85, 90, 95]，因为它们指向同一块内存

print("\n" + "-" * 20, "2. 浅拷贝 copy.copy()", "-" * 20)
# 重新新建原始对象
orig_2 = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(orig_2)  # 浅拷贝
shallow["scores"].append(95) # 修改内部列表
print("原对象 (orig_2):", orig_2)
print("浅拷贝 (shallow):", shallow)
# 结果：原对象内部的列表也会变成 [85, 90, 95]！
# 原因：浅拷贝只复制了最外层的字典，内部的列表依然是共享的（指向同一块内存）

print("\n" + "-" * 20, "3. 深拷贝 copy.deepcopy()", "-" * 20)
# 重新新建原始对象
orig_3 = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(orig_3) # 深拷贝
deep["scores"].append(95)    # 修改内部列表
print("原对象 (orig_3):", orig_3)
print("深拷贝 (deep)  :", deep)
# 结果：原对象内部的列表保持 [85, 90] 不变！
# 原因：深拷贝连同内部的列表也彻底复制了一份，两者完全独立
```
