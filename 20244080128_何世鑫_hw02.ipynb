# 字符串的乐趣

```python
print("1")
def is_palindrome(s):
    t = "".join(ch for ch in s.lower() if ch != " ")
    return t == t[::-1]
```

```python
print("2")
def is_abecedarian(s):
    t = "".join(ch for ch in s if ch != " ")
    for i in range(len(t) - 1):
        if t[i] > t[i + 1]:
            return False
    return True
```

```python
print("3")
def remove_vowels(s):
    vowels = set("aeiou")
    return "".join(ch for ch in s if ch not in vowels)
```

```python
print(is_palindrome("level"))            # True
print(is_palindrome("taco cat"))         # True
print(is_palindrome("T A C O cat"))      # True
print(is_palindrome("hello"))            # False

print(is_abecedarian("adder"))           # True
print(is_abecedarian("beet"))            # True
print(is_abecedarian("dog"))             # False
print(is_abecedarian("abcd efgh xyz"))   # True

print(remove_vowels("cat"))              # ct
print(remove_vowels("audio"))            # d
print(remove_vowels("aeiou"))            # '' (空串)
```

# 列表、字典和类

```python
print("1")
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("list_reverse expects a list")
    return lst[::-1]
```

```python
print("2")
def is_sorted(seq):
    it = iter(seq)
    try:
        prev = next(it)
    except StopIteration:
        return True
    for x in it:
        if x < prev:
            return False
        prev = x
    return True
```

```python
print(list_reverse([1, 2, 3]))      # [3, 2, 1]
print(list_reverse([]))             # []
try:
    list_reverse("abc")
except TypeError as e:
    print("TypeError:", e)

print(is_sorted([1, 2, 3]))         # True
print(is_sorted([1, 3, 2]))         # False
print(is_sorted([]))                # True
print(is_sorted([5]))               # True
print(is_sorted("abc"))             # True
print(is_sorted((3, 2, 1)))         # False
```

# 作为向量的元组

```python
print("1")
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("t must be a tuple")
    if not isinstance(s, (int, float)) or isinstance(s, bool):
        raise TypeError("s must be an int or float")
    if not all(isinstance(x, (int, float)) and not isinstance(x, bool) for x in t):
        raise TypeError("tuple elements must be int or float")
    return tuple(x * s for x in t)
```

```python
print("2")
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("both inputs must be tuples")
    if len(x) != len(y):
        raise ValueError("vector dimensions must match")
    return float(sum(a * b for a, b in zip(x, y)))
```

```python
print("3")
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True          # 空矩阵视为合法（0 行）
    if not all(isinstance(row, tuple) for row in mx):
        return False
    n = len(mx[0])
    for row in mx:
        if len(row) != n:
            return False
        if not all(isinstance(x, (int, float)) and not isinstance(x, bool) for x in row):
            return False
    return True
```

```python
print("4")
def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise ValueError("mx is not a valid matrix")
    if not isinstance(v, tuple):
        raise TypeError("v must be a tuple")
    if not all(isinstance(x, (int, float)) and not isinstance(x, bool) for x in v):
        raise TypeError("vector elements must be int or float")
    n_cols = len(mx[0]) if len(mx) > 0 else 0
    if len(v) != n_cols:
        raise ValueError("vector length must equal matrix column count")
    return tuple(
        float(sum(mx[i][j] * v[j] for j in range(n_cols)))
        for i in range(len(mx))
    )
```

```python
# 1
print(vec_scalar_mult((1, 2, 3), 2))        # (2, 4, 6)
print(vec_scalar_mult((1.5, 2), 2))         # (3.0, 4)
print(vec_scalar_mult((), 5))               # ()

# 2
print(vec_inner_product((1, 2, 3), (4, 5, 6)))  # 32.0
print(vec_inner_product((), ()))                # 0.0
try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)

# 3
print(check_valid_mx(((1, 2), (3, 4))))     # True
print(check_valid_mx(((1, 2), (3,))))       # False
print(check_valid_mx(([1, 2], [3, 4])))     # False（行是 list）
print(check_valid_mx(()))                   # True

# 4
print(mx_vec_mult(((1, 2), (3, 4)), (1, 1)))   # (3.0, 7.0)
print(mx_vec_mult((), ()))                      # ()
```

# 更多向量的乐趣（面向对象）

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError("dim must be an int")
        if dim < 0:
            raise ValueError("dim must be non-negative")
        if elements is None:
            elements = (0,) * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("elements must be a list or tuple")
        if len(elements) != dim:
            raise ValueError("dim and number of elements must match")
        self.dim = dim
        self.elements = tuple(elements)

    # 打印对象时显示友好格式
    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    # 相等比较
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("cannot compare vectors of different dimensions")
        return self.elements == other.elements

    # 小于比较
    def __lt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("cannot compare vectors of different dimensions")
        return self.elements < other.elements

    # 小于等于比较
    def __le__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("cannot compare vectors of different dimensions")
        return self.elements <= other.elements

    # 大于比较
    def __gt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("cannot compare vectors of different dimensions")
        return self.elements > other.elements

    # 点乘（内积）
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot expects a Vector")
        if self.dim != other.dim:
            raise ValueError("vector dimensions must match")
        return float(sum(a * b for a, b in zip(self.elements, other.elements)))

    # 乘法（支持 v * 2，v * w）
    def __mul__(self, other):
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return Vector(self.dim, tuple(x * other for x in self.elements))
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("vector dimensions must match")
            return Vector(self.dim, tuple(a * b for a, b in zip(self.elements, other.elements)))
        return NotImplemented

    # 乘法（支持 2 * v）
    def __rmul__(self, other):
        return self.__mul__(other)


# ====== 下面是测试代码，也会一起运行 ======
v = Vector(2, (2, 3))
w = Vector(2, (4, 5))

print(v)                # 应该输出：Vector(2, (2, 3))
print(v == Vector(2, (2, 3)))   # 应该输出：True
print(v < w)            # 应该输出：True
print(v <= w)           # 应该输出：True
print(w > v)            # 应该输出：True

print(v.dot(w))         # 应该输出：23.0
print(v * 2)            # 应该输出：Vector(2, (4, 6))
print(2 * v)            # 应该输出：Vector(2, (4, 6))
print(v * w)            # 应该输出：Vector(2, (8, 15))

print(Vector(3))        # 应该输出：Vector(3, (0, 0, 0))
```
